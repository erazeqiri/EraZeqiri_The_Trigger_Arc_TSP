```python

from pathlib import Path
import pandas as pd

file_path = Path("data/example.txt")

with open(file_path, "r", encoding="utf-8") as file:
    lines = file.read().splitlines()

n, m, r = map(int, lines[0].split())

print("Numri i nyjeve:", n)
print("Numri i harqeve:", m)
print("Numri i relacioneve trigger:", r)
```

```python

# Lexojme harqet nga example.txt
arcs_data = []

for line in lines[1:1 + m]:
    arc_id, from_node, to_node, cost = line.split()

    arcs_data.append({
        "Arc ID": int(arc_id),
        "From": int(from_node),
        "To": int(to_node),
        "Cost": float(cost)
    })

# Krijojme tabelen me pandas
df_arcs = pd.DataFrame(arcs_data)

display(df_arcs)
```
