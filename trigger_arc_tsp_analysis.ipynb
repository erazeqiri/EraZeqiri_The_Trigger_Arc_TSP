```python
def improved_initial_route(
    n, arcs, max_steps=100000, max_restarts=10
):
    adjacency = {node: set() for node in range(n)}

    for arc in arcs:
        adjacency[arc["from"]].add(arc["to"])

    for attempt in range(max_restarts):

        route = [0]
        visited = {0}
        steps = 0

        def backtrack():
            nonlocal steps

            if steps >= max_steps:
                return False

            steps += 1

            if len(route) == n:
                return 0 in adjacency[route[-1]]

            current = route[-1]

            neighbors = [
                node for node in adjacency[current]
                if node not in visited
            ]

            random.shuffle(neighbors)

            neighbors.sort(
                key=lambda node: len(
                    adjacency[node] - visited - {node}
                )
            )

            for node in neighbors:

                route.append(node)
                visited.add(node)

                if backtrack():
                    return True

                route.pop()
                visited.remove(node)

                if steps >= max_steps:
                    break

            return False

        if backtrack():
            return route + [0]

    return None
```

```python
from pathlib import Path

source = Path(
    "/kaggle/working/EraZeqiri_The_Trigger_Arc_TSP/src/simulated_annealing.py"
)

code = source.read_text(encoding="utf-8")

start = code.index("def generate_neighbor(")
end = code.index("\ndef simulated_annealing(", start)

new_function = '''
def generate_neighbor(route, arcs=None, max_attempts=100):
    if len(route) < 4:
        return route.copy()

    existing_arcs = None
    if arcs is not None:
        existing_arcs = {
            (arc["from"], arc["to"]) for arc in arcs
        }

    for _ in range(max_attempts):
        neighbor = route.copy()
        operation = random.choice(["swap", "relocate", "segment"])

        if operation == "swap":
            i, j = random.sample(range(1, len(neighbor) - 1), 2)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

        elif operation == "relocate":
            i, j = random.sample(range(1, len(neighbor) - 1), 2)
            node = neighbor.pop(i)
            neighbor.insert(j, node)

        else:
            internal = neighbor[1:-1]

            if len(internal) < 3:
                continue

            start = random.randrange(0, len(internal) - 1)
            end = random.randrange(start + 2, len(internal) + 1)

            segment = internal[start:end]
            remaining = internal[:start] + internal[end:]

            insert_at = random.randrange(len(remaining) + 1)

            new_internal = (
                remaining[:insert_at]
                + segment
                + remaining[insert_at:]
            )

            neighbor = [route[0]] + new_internal + [route[-1]]

        if neighbor == route:
            continue

        if existing_arcs is None or all(
            (neighbor[k], neighbor[k + 1]) in existing_arcs
            for k in range(len(neighbor) - 1)
        ):
            return neighbor

    return route.copy()
'''

updated = code[:start] + new_function + code[end:]

compile(updated, str(source), "exec")
source.write_text(updated, encoding="utf-8")

print("Segment Relocation u integrua ne Kaggle!")
print("Skedari:", source)
```

```python
import ast
import pandas as pd

from cost_calculator import CostCalculator
from validator import is_valid_route
from parser import read_instance

df = pd.read_csv(
    "/kaggle/working/best_55_with_segment.csv"
)

paths = {p.name: p for p in all_instances}
errors = []

for _, row in df.iterrows():
    path = paths[row["Instance"]]
    n, m, r, arcs, relations = read_instance(path)

    route = ast.literal_eval(row["Route"])
    valid = is_valid_route(route, n, arcs)
    actual_cost = CostCalculator(arcs, relations).calculate(route)

    if (
        not valid
        or actual_cost is None
        or abs(actual_cost - row["Cost"]) > 0.011
    ):
        errors.append(row["Instance"])

print("Instanca te verifikuara:", len(df))
print("Gabime:", len(errors))
print("Kosto totale:", round(df["Cost"].sum(), 2))

if not errors:
    print("Te gjitha rruget dhe kostot u verifikuan!")
else:
    print("Instancat me gabime:", errors)
```

```python

import pandas as pd
from pathlib import Path

# Gjetja e skedareve
input_path = Path("/kaggle/input")

baseline_path = next(
    input_path.rglob("baseline_results.csv")
)

final_path = next(
    input_path.rglob("best_55_trigger_guided.csv")
)

# Leximi i rezultateve
baseline = pd.read_csv(baseline_path)
final = pd.read_csv(final_path)

# Kontrolli i instancave
assert len(baseline) == 55
assert len(final) == 55
assert baseline["Instance"].is_unique
assert final["Instance"].is_unique
assert set(baseline["Instance"]) == set(final["Instance"])

# Bashkimi i rezultateve
comparison = baseline[["Instance", "Cost", "Time"]].merge(
    final[["Instance", "Cost", "Valid"]],
    on="Instance",
    suffixes=("_Baseline", "_Final"),
    validate="one_to_one"
)

comparison = comparison.rename(
    columns={"Time": "Baseline_Time"}
)

# Llogaritja e permiresimit
comparison["Improvement"] = (
    comparison["Cost_Baseline"] -
    comparison["Cost_Final"]
)

comparison["Improvement_Percent"] = (
    comparison["Improvement"] /
    comparison["Cost_Baseline"] * 100
)

# Statistikat perfundimtare
baseline_total = comparison["Cost_Baseline"].sum()
final_total = comparison["Cost_Final"].sum()

improvement = baseline_total - final_total
percentage = improvement / baseline_total * 100

valid_count = (
    comparison["Valid"].astype(str).str.lower() == "true"
).sum()

improved_count = (
    comparison["Improvement"] > 1e-8
).sum()

equal_count = (
    comparison["Improvement"].abs() <= 1e-8
).sum()

worse_count = (
    comparison["Improvement"] < -1e-8
).sum()

# Paraqitja e rezultateve
print("=" * 55)
print("THE TRIGGER ARC TSP - FINAL RESULTS")
print("=" * 55)

print(f"Total instances: {len(comparison)}")
print(f"Saved valid routes: {valid_count}/55")
print(f"Baseline total cost: {baseline_total:.2f}")
print(f"Final total cost: {final_total:.2f}")
print(f"Total improvement: {improvement:.2f}")
print(f"Improvement percentage: {percentage:.2f}%")
print(f"Improved instances: {improved_count}")
print(f"Unchanged instances: {equal_count}")
print(f"Worse instances: {worse_count}")

print("=" * 55)

# Tabela krahasuese
display(comparison.round(4))

# Ruajtja e analizes
output_path = "/kaggle/working/final_comparison.csv"

comparison.to_csv(output_path, index=False)

print(f"Results saved to: {output_path}")

```
