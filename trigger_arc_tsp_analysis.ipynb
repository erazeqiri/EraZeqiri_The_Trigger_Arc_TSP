```python

from pathlib import Path
import pandas as pd

file_path = Path("data/example.txt")

with open(file_path, "r", encoding="utf-8") as file:
    lines = file.read().splitlines()

n, m, r = map(int, lines[0].split())

print("Numri i nyjeve:", n)
print("Numri i harqeve:", m)
print("Numri i relacioneve trigger:", r)
```

```python

# Lexojme harqet nga example.txt
arcs_data = []

for line in lines[1:1 + m]:
    arc_id, from_node, to_node, cost = line.split()

    arcs_data.append({
        "Arc ID": int(arc_id),
        "From": int(from_node),
        "To": int(to_node),
        "Cost": float(cost)
    })

# Krijojme tabelen me pandas
df_arcs = pd.DataFrame(arcs_data)

display(df_arcs)
```

```python

import matplotlib.pyplot as plt

# Pozicionet e nyjeve
positions = {
    0: (0, 1),
    1: (4, 3),
    2: (2, 3),
    3: (2, 1),
    4: (4, 0)
}

fig, ax = plt.subplots(figsize=(10, 7))

# Vizatojme harqet dhe kostot
for _, arc in df_arcs.iterrows():
    start = int(arc["From"])
    end = int(arc["To"])
    cost = arc["Cost"]

    x1, y1 = positions[start]
    x2, y2 = positions[end]

    ax.annotate(
        "",
        xy=(x2, y2),
        xytext=(x1, y1),
        arrowprops=dict(
            arrowstyle="->",
            color="gray",
            lw=1.8,
            shrinkA=22,
            shrinkB=22,
            connectionstyle="arc3,rad=0.12"
        )
    )

    dx = x2 - x1
    dy = y2 - y1

    ax.text(
        (x1 + x2) / 2 + 0.09 * dy,
        (y1 + y2) / 2 - 0.09 * dx,
        f"{cost:g}",
        color="black",
        fontsize=11,
        ha="center",
        bbox=dict(facecolor="white", edgecolor="none")
    )

# Vizatojme nyjet
for node, (x, y) in positions.items():
    color = "red" if node == 0 else "dodgerblue"

    ax.scatter(x, y, s=1400, color=color, zorder=3)

    ax.text(
        x, y, str(node),
        color="white",
        fontsize=15,
        fontweight="bold",
        ha="center",
        va="center",
        zorder=4
    )

ax.set_title("The Trigger Arc TSP - Directed Graph")
ax.set_xlim(-0.7, 4.7)
ax.set_ylim(-0.7, 3.7)
ax.set_aspect("equal")
ax.axis("off")

plt.show()
```

```python

# Lexojme relacionet trigger
relations_data = []

for line in lines[1 + m:1 + m + r]:
    values = line.split()

    relations_data.append({
        "Relation ID": int(values[0]),
        "Trigger Arc": int(values[1]),
        "Trigger From": int(values[2]),
        "Trigger To": int(values[3]),
        "Target Arc": int(values[4]),
        "Target From": int(values[5]),
        "Target To": int(values[6]),
        "New Cost": float(values[7])
    })

# Krijojme tabelen e relacioneve
df_relations = pd.DataFrame(relations_data)

display(df_relations)
```
